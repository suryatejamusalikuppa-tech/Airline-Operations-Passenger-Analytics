Create suitable age groups.

# SECTION 6 — SATISFACTION ANALYSIS

## Q21. Overall Passenger Satisfaction

Find:

- Satisfied passengers

## Q23. Satisfaction by Customer Type

```python
satisfaction_by_customer = df.groupby('Customer Type')['satisfaction'].apply(lambda x: (x == 'satisfied').mean() * 100)
print("Satisfaction percentage by customer type:\n", satisfaction_by_customer)
```

## Q25. Delay vs Passenger Satisfaction

Create a new column:

`Departure_Delay_Status`

with:

- Delayed → Departure Delay > 0
- Not Delayed → Departure Delay = 0

Compare passenger satisfaction between the two groups.

### Business Question

**Does passenger satisfaction differ between delayed and non-delayed flights in this dataset?**



```python
df['Departure_Delay_Status'] = np.where(df['Departure Delay in Minutes'] > 0, 'Delayed', 'Not Delayed')
satisfaction_by_delay = df.groupby('Departure_Delay_Status')['satisfaction'].apply(lambda x: (x == 'satisfied').mean() * 100)
print("Satisfaction percentage by departure delay status:\n", satisfaction_by_delay)
```
